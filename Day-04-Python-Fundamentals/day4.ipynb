```python
#3. Practice try, except, and finally.
#A. try and except
try:
    a = 10
    b = 0

    result = a/b

    print(result)

except ZeroDivisionError:
    print("Cannot divide a number by zero")
```

```python
#B. Handling ValueError
try:
    age=int(input("Enter your age:"))

    print("Your age is:")

except ValueError:
    print("Please enter a valid number")
```

```python
#C. Handling FileNotFoundError
try:
    file = open("data.txt","r")
    print(file.read())
    file.close()

except FileNotFoundError:
    print("File does not exist")
```

```python
#D. finally
try:
    num = 10/2

    print(num)

except ZeroDivisionError:
    print("Cannot divide by zero")

finally:
    print("Program Execution completed")
```

```python
#E. else
try:
    num = 10/2

except ZeroDivisionError:
    print("Error occurred")

else:
    print("Result:",num)

finally:
    print("Task completed")
```

```python
#5. Create a program that reads a CSV file, checks whether it exists, and handles invalid data. 
#Step 1: Create a CSV file
import csv

with open("Student_data.csv","w",newline="") as file:
    writer = csv.writer(file)

    writer.writerow(["Name","Age","Marks"])

    writer.writerow(["Sanika",23,89])
    writer.writerow(["Ajay",23,85])
    writer.writerow(["Sam",21,87])
    writer.writerow(["Sandy",20,78])
    writer.writerow(["Chaitrali",30,88])
    writer.writerow(["Anit",22,70])
    writer.writerow(["Ritu",20,75])
    writer.writerow(["Sanu",18,80])
```

```python
#Step 2:
import csv
import os

filename = "Student_data.csv"

try:
    if not os.path.exists(filename):
        raise FileNotFoundError("CSV file not found")

    with open(filename, "r") as file:
        reader = csv.DictReader(file)

        print("Valid student records:")
        print("----------------------")

        for row in reader:

            try:

                name = row["Name"].strip()
                age = int(row["Age"])
                marks = float(row["Marks"])

                if not name:
                    raise ValueError("Name is missing")

                if age <= 0:
                    raise ValueError("Invalid age")

                if marks < 0 or marks > 100:
                    raise ValueError("Marks must be between 0 and 100")

                print(name, age, marks)

            except ValueError as e:
                print("Invalid record:", row, "/ Reason:",e)

except FileNotFoundError:
    print("Error: CSV file does not exist")

except PermissionError:
    print("Error: You do not have permission to read this file")

except Exception as e:
    print("Unexpected error:",e)

finally:
    print("CSV file processing completed")
```
