```python
#5. Compare basic Python calculations with NumPy operations. 
#1: Addition
no=[10,20,30]
result=[x + 5 for x in no]
print(result)

#Using NumPy
import numpy as np
no=np.array([10,20,30])
result = no + 5
print(result)
```

```python
#2: Multiplication
no=[10,20,30]
result = [x * 2 for x in no]
print(result)

#Using NumPy
no=np.array([10,20,30])
result=no*2
print(result)
```

```python
#3: Average
marks=[70,80,90,100]
average=sum(marks)/len(marks)
print(average)

#Using NumPy
marks=np.array([70,80,90,100])
print(np.mean(marks))
```

```python
#Q13. Calculate the percentage of students who scored above 60.
marks = np.array([45, 65, 80, 55, 90, 75, 40, 85])

passed = marks[marks > 60]
percentage = (len(passed) / len(marks))*100

print("Passed student:",len(passed))
print("Pass percentage:",percentage)
```
