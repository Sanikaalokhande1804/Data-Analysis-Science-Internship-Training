```python
#7. Practical Programs for Data Analytics
name = input("Enter student name: ")

maths = float(input("Enter Maths marks: "))
science = float(input("Enter Science marks: "))
english = float(input("Enter English marks: "))
python = float(input("Enter Python marks: "))
sql = float(input("Enter SQL marks: "))

total = maths + science + english + python + sql

percentage = (total / 500) * 100

print("\nStudent Name:", name)
print("Total Marks:", total)
print("Percentage:", percentage, "%")
```

```python
product1 = float(input("Enter product 1 price: "))
quantity1 = int(input("Enter product 1 quantity: "))

product2 = float(input("Enter product 2 price: "))
quantity2 = int(input("Enter product 2 quantity: "))

product3 = float(input("Enter product 3 price: "))
quantity3 = int(input("Enter product 3 quantity: "))

sales1 = product1 * quantity1
sales2 = product2 * quantity2
sales3 = product3 * quantity3

total_sales = sales1 + sales2 + sales3

print("Product 1 Sales:", sales1)
print("Product 2 Sales:", sales2)
print("Product 3 Sales:", sales3)

print("Total Sales:", total_sales)
```

```python
price = float(input("Enter product price: "))
discount_percent = float(input("Enter discount percentage: "))

discount = (price * discount_percent) / 100

final_price = price - discount

print("Original Price:", price)
print("Discount Amount:", discount)
print("Final Price:", final_price)
```
