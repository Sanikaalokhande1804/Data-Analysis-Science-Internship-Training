```python
#Q19. Create a function to count even numbers in a list.
def count_even(numbers):
    count = 0

    for num in numbers:
        if num % 2 == 0:
            count += 1

    return count

numbers = [10,21,30,45,30,15,10]
print("Even Count:",count_even(numbers))
```
